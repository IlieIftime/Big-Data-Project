#### Obter os dados

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import requests
from pymongo import MongoClient
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, trim, count, when, isnan, isnull, avg, last, floor, lit
from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.types import StringType
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation
from pyspark.sql.types import DoubleType, IntegerType
import pycountry_convert as pc
```

```python
spark = SparkSession.builder.appName("SparkMongoDBAnalysis").getOrCreate()
```

```python
#Lê os dados de eletricidade a partir do MongoDB
df_eletricidade_raw = spark.read \
    .format("mongodb") \
    .option("connection.uri", "mongodb+srv://ilie:ilieiftime@cluster0.xrpysci.mongodb.net/?retryWrites=true&w=majority") \
    .option("database", "ts8") \
    .option("collection", "elec") \
    .load()

#Mostra a estrutura da coleção de eletricidade e mostra as 5 primeiras linhas
df_eletricidade_raw.printSchema()
df_eletricidade_raw.show(5)


#Lê os dados de CO₂ a partir do MongoDB
df_co2_raw = spark.read \
    .format("mongodb") \
    .option("connection.uri", "mongodb+srv://ilie:ilieiftime@cluster0.xrpysci.mongodb.net/?retryWrites=true&w=majority") \
    .option("database", "ts8") \
    .option("collection", "co2") \
    .load()

#Mostra a estrutura da coleção de CO2 e mostra as 5 primeiras linhas
df_co2_raw.printSchema()
df_co2_raw.show(5)
```

#### Limpeza e pré-processamento

```python
#Renomear para nomes simples e em português
df_co2 = df_co2_raw.withColumnRenamed("Entity", "País") \
                   .withColumnRenamed("Year", "Ano") \
                   .withColumnRenamed("Code", "Código_país_co2") \
                   .withColumnRenamed("_id","ID_co2") \
                   .withColumnRenamed("Annual CO2 emissions (per capita)", "Valor_CO2")

df_ele = df_eletricidade_raw.withColumnRenamed("Entity", "País") \
                            .withColumnRenamed("Year", "Ano") \
                            .withColumnRenamed("Code", "Código_país_ele") \
                            .withColumnRenamed("_id","ID_ele") \
                            .withColumnRenamed("Access to electricity (% of population)", "Valor_Elec")

#Mostra a estrutura das coleções renomeadas e mostra as 10 primeiras linhas
df_co2.printSchema()
df_ele.printSchema()

df_co2.show(10)
df_ele.show(10)
```

```python
#Calcula o número de valores nulos em cada coluna do dataset de CO₂
nulos_por_coluna_co2 = df_co2.select([
    F.sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in df_co2.columns
])
nulos_por_coluna_co2.show()

#Calcula o número de valores nulos em cada coluna do dataset de Eletricidade
nulos_por_coluna_ele = df_ele.select([
    F.sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in df_ele.columns
])
nulos_por_coluna_ele.show()

#Soma todos os valores nulos dos datasets
nulos_totais_co2 = nulos_por_coluna_co2.toPandas().sum().sum()
nulos_totais_ele = nulos_por_coluna_ele.toPandas().sum().sum()

print("Nulos totais CO2:", nulos_totais_co2)
print("Nulos totais Eletricidade:", nulos_totais_ele)
```

```python
#Filtra os datasets para incluir apenas os anos entre 2000 e 2020
df_co2 = df_co2.filter((col("Ano") >= 2000) & (col("Ano") <= 2020))
df_ele = df_ele.filter((col("Ano") >= 2000) & (col("Ano") <= 2020))

df_co2.select("Ano").describe().show()
df_ele.select("Ano").describe().show()
```

#### Integração dos dados

```python
#Junta os dois datasets (CO₂ e Eletricidade) usando País e Ano como chave
df_final = df_co2.join(
    df_ele,
    on=["País", "Ano"],
    how="inner"
)

df_final.show(10)
```

```python
#Seleciona apenas as colunas necessárias do DataFrame final
df_final = df_final.select(
    "País",
    "Ano",
    "Código_país_co2",
    "Valor_CO2",
    "Valor_Elec"
)

df_final.show(10)
```

#### Análise Relacional

```python
#Converte o DataFrame Spark para Pandas
try:
    from pyspark.sql import DataFrame as SparkDF
    if isinstance(df_final, SparkDF):
        df_pd = df_final.toPandas()
    else:
        df_pd = df_final.copy()
except:
    df_pd = df_final.copy()

#Certifica que os valores são numéricos
df_pd["Valor_CO2"] = pd.to_numeric(df_pd["Valor_CO2"], errors="coerce")
df_pd["Valor_Elec"] = pd.to_numeric(df_pd["Valor_Elec"], errors="coerce")

#Calcula a média por país
df_paises = df_pd.groupby("País")[["Valor_CO2", "Valor_Elec"]].mean()

#Calcula a correlação entre as duas variáveis
corr_matrix = df_paises.corr()
print("Correlação entre CO₂ per capita e Acesso à Eletricidade:")
print(corr_matrix)

#Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlação (por país)\nCO₂ per capita vs Acesso à Eletricidade")
plt.show()

#Gráafico de dipersão (cada ponto é um país)
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df_paises,
    x="Valor_Elec",
    y="Valor_CO2"
)

plt.title("Relação entre CO₂ per capita e Acesso à Eletricidade\n(diferentes países)")
plt.xlabel("Acesso à Eletricidade (%)")
plt.ylabel("CO₂ per capita")
plt.grid(True)
plt.show()
```

```python
#Converte o DataFrame Spark para Pandas
try:
    from pyspark.sql import DataFrame as SparkDF
    if isinstance(df_final, SparkDF):
        df_pd = df_final.toPandas()
    else:
        df_pd = df_final.copy()
except:
    df_pd = df_final.copy()

#Certifica que os valores são numéricos
df_pd["Ano"] = pd.to_numeric(df_pd["Ano"], errors="coerce")
df_pd["Valor_CO2"] = pd.to_numeric(df_pd["Valor_CO2"], errors="coerce")
df_pd["Valor_Elec"] = pd.to_numeric(df_pd["Valor_Elec"], errors="coerce")

#Tendência temporal de CO2
plt.figure(figsize=(10,5))
sns.lineplot(data=df_pd, x="Ano", y="Valor_CO2", estimator="mean")
plt.title("Tendência Global das Emissões de CO₂ per capita")
plt.ylabel("CO₂ per capita (média global)")
plt.xlabel("Ano")
plt.show()

#Tendência temporal de eletricidade
plt.figure(figsize=(10,5))
sns.lineplot(data=df_pd, x="Ano", y="Valor_Elec", estimator="mean")
plt.title("Tendência Global do Acesso à Eletricidade (%)")
plt.ylabel("Acesso à Eletricidade (%) (média global)")
plt.xlabel("Ano")
plt.show()

#Tendência de CO2 por país
plt.figure(figsize=(12,6))
sns.lineplot(data=df_pd, x="Ano", y="Valor_CO2", hue="País", estimator="mean")
plt.title("Tendência das Emissões de CO₂ per capita por País")
plt.ylabel("CO₂ per capita (média por país)")
plt.xlabel("Ano")
plt.legend(title="País", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

#Tendência de eletricidade por país
plt.figure(figsize=(12,6))
sns.lineplot(data=df_pd, x="Ano", y="Valor_Elec", hue="País", estimator="mean")
plt.title("Tendência do Acesso à Eletricidade por País")
plt.ylabel("Acesso à Eletricidade (%) (média por país)")
plt.xlabel("Ano")
plt.legend(title="País", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

#### Discussão dos Resultados

A análise mostra que existe uma relação positiva entre as emissões de CO₂ per capita e o acesso à eletricidade nos diferentes países. A correlação (~0.47) indica que, de forma geral, países com maior acesso à eletricidade tendem também a apresentar valores mais elevados de emissões per capita, o que está associado ao nível de desenvolvimento económico e ao maior consumo energético.

O gráfico de dispersão reforça esta ideia:
A maioria dos países com eletrificação acima de 90% apresenta emissões superiores, enquanto países com baixo acesso à eletricidade concentram-se em níveis muito reduzidos de CO₂. Também se observam alguns países com emissões extremamente altas, situando-se como outliers.

Nas tendências globais (2000–2020), verifica-se que o acesso médio à eletricidade tem aumentado de forma consistente ao longo dos anos, refletindo progressos na expansão energética mundial. Em contraste, as emissões médias de CO₂ exibem maior variação, com ligeiro crescimento até cerca de 2010 e redução gradual nos últimos anos, sugerindo mudanças nos padrões de consumo e maior adoção de energias mais limpas.

Ao analisar as tendências por país, nota-se grande heterogeneidade. Alguns países apresentam trajetórias de emissões muito elevadas e instáveis, enquanto outros mantêm valores baixos ou apresentam melhorias. O acesso à eletricidade também evolui de forma desigual: apesar de muitos países atingirem quase universalização, vários ainda apresentam progressos graduais ou níveis reduzidos.
