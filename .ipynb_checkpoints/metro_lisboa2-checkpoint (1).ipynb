Em df_mob, vemos dados de estações GIRA, como station_id, num_bicicletas, num_docas e update_date. Os dados estão no seu formato bruto, prontos para a próxima fase de limpeza e transformação.

## Módulo 2 — Limpeza e Transformação dos Dados

Os dados brutos raramente estão prontos para análise. Este módulo foca-se em limpar e transformar os DataFrames. As principais tarefas incluem:
Separar o DataFrame df_mob em gira, que contém os dados de disponibilidade.
Converter campos de data/hora de strings para o tipo timestamp, que é otimizado para operações temporais.
Normalizar texto, como descrições de trânsito, para garantir consistência (ex: convertendo para minúsculas).
Filtrar valores inválidos ou nulos para garantir a qualidade dos dados.
O resultado final são dois DataFrames limpos e estruturados: gira_clean e transito_clean.


```python
# Imports necessários para este módulo
from pyspark.sql import functions as F
from pyspark.sql.functions import col, udf, from_json
from pyspark.sql.types import StructType, StructField, StringType, ArrayType, DoubleType
import re
import unicodedata
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

def remover_acentos(texto):
    if texto is None:
        return None
    texto = unicodedata.normalize("NFKD", texto)
    texto = texto.encode("ASCII", "ignore").decode("ASCII")
    return texto

remover_acentos_udf = udf(remover_acentos, StringType())

# 1. Criar DataFrame gira e transito
gira = df_mob
transito = df_cond

# 2. Limpeza e transformação para gira_clean
gira_clean = gira.withColumn("update_date_ts", F.to_timestamp(col("entity_ts"))) \
                   .filter(col("numbicicletas").isNotNull() & col("numdocasvacias").isNotNull()) \
                   .withColumn("estacao_normalizada", remover_acentos_udf(col("desigcomercial"))) \
                   .withColumn("data", F.to_date(col("entity_ts"))) \
                   .select(
                       col("desigcomercial").alias("station_name"),
                       col("numbicicletas").alias("num_bicicletas"),
                       col("numdocasvacias").alias("num_docas_vacias"),
                       col("estacao_normalizada").alias("estacao_normalizada"),
                       "update_date_ts",
                       "data"
                   )

# 3. Limpeza e transformação para transito_clean

# UDF para normalização de texto
def normalize_text(text):
    if text is None:
        return None
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

normalize_text_udf = udf(normalize_text, StringType())

# Passo C: Construir o DataFrame final usando a nova coluna 'position_struct' e a coluna 'motivo'
from pyspark.sql.functions import col, get_json_object, to_timestamp

transito_clean = (
    df_cond
    # timestamp
    .withColumn("creation_date_ts", to_timestamp(col("creation_date")))
    .withColumn("data", F.to_date(col("creation_date_ts")))

    # extrair latitude e longitude do GeometryCollection
    .withColumn(
        "lon",
        get_json_object(col("position"), "$.geometries[0].coordinates[0][0][0]").cast("double")
    )
    .withColumn(
        "lat",
        get_json_object(col("position"), "$.geometries[0].coordinates[0][0][1]").cast("double")
    )

    # normalizar descrição
    .withColumn("descricao_normalizada", normalize_text_udf(col("motivo")))

    # filtrar apenas registos válidos
    .filter(col("lat").isNotNull() & col("lon").isNotNull())

    # selecionar colunas finais
    .select(
        "entity_id",
        "creation_date_ts",
        "descricao_normalizada",
        "lat",
        "lon",
        "data"
    )
)

print("Schema de gira_clean:")
gira_clean.printSchema()
gira_clean.show(5, truncate=False)

print("\nSchema de transito_clean:")
transito_clean.printSchema()
transito_clean.show(5, truncate=False)
```


Concluímos a etapa de transformação com sucesso.
gira_clean: Contém agora uma coluna update_date_ts do tipo timestamp e uma coluna data apenas com a data, facilitando agregações diárias. Linhas com valores nulos em colunas críticas foram removidas.
transito_clean: Possui uma coluna de timestamp (creation_date_ts), uma coluna de data (data) e uma nova coluna descricao_normalizada com o texto limpo pela UDF. Registos sem coordenadas foram descartados para garantir a integridade dos dados geográficos.
Ambos os DataFrames estão agora robustos, limpos e prontos para as fases de agregação e análise.

## Módulo 3 — Agregações Diárias e Integração de Datasets
Explicação Curta
Com os dados limpos, podemos começar a extrair valor através de agregações. O objetivo deste módulo é resumir os dados a um nível diário.
gira_por_dia: Calculamos a média de bicicletas disponíveis em todas as estações para cada dia.
transito_por_dia: Contamos o número total de condicionamentos de trânsito reportados a cada dia.
df_integrado: Unimos os dois DataFrames agregados pela coluna data. Este DataFrame final permitirá analisar a correlação entre a disponibilidade de bicicletas e os incidentes de trânsito.

```python
# 1. Agregar dados da GIRA por dia (média de bicicletas)
from pyspark.sql.functions import round

gira_por_dia = (
    gira_clean
    .groupBy("data")
    .agg(round(avg("num_bicicletas"), 2).alias("media_bicicletas"))
    .orderBy("data")
)

print("Agregação - Média de bicicletas por dia:")
gira_por_dia.show(5)

# 2. Agregar dados de trânsito por dia (número de condicionamentos)
transito_por_dia = transito_clean.groupBy("data") \
                                 .agg(count("*").alias("num_condicionamentos")) \
                                 .orderBy("data")

print("\nAgregação - Número de condicionamentos por dia:")
transito_por_dia.show(5)

# 3. Integrar os dois datasets
df_integrado = gira_por_dia.join(transito_por_dia, on="data", how="inner").orderBy("data")

print("\nDataFrame Integrado:")
df_integrado.show(10)
```


Os resultados mostram os DataFrames agregados e o DataFrame final integrado.
gira_por_dia e transito_por_dia resumem eficazmente as métricas chave (média de bicicletas e contagem de incidentes) a nível diário.
df_integrado combina estas duas métricas numa única tabela. A utilização de um inner join garante que apenas analisamos os dias para os quais temos dados de ambas as fontes. Este DataFrame é a base para a nossa análise exploratória de séries temporais.

## Módulo 7 — Modelos Preditivos com PySpark ML
Explicação Curta
Após explorar e visualizar os dados, o próximo passo lógico é utilizar Machine Learning para criar modelos preditivos. Esta abordagem proativa permite antecipar eventos futuros com base em padrões históricos. Neste módulo, desenvolveremos dois modelos distintos:
Modelo de Regressão: Para prever o número de bicicletas disponíveis numa estação específica e num determinado momento. Isto pode otimizar a logística de redistribuição.
Modelo de Classificação: Para prever a probabilidade de congestionamento numa zona e hora específicas. Isto pode ser usado para sistemas de alerta de trânsito ou planeamento urbano.
Utilizaremos a biblioteca pyspark.ml para construir pipelines de Machine Learning robustos e escaláveis.
7.1 Previsão de Disponibilidade de Bicicletas (Regressão)
O objetivo aqui é prever uma variável contínua: num_bicicletas. Este é um problema clássico de regressão. Usaremos características (features) como a estação, a hora do dia e o dia da semana para treinar um modelo Gradient Boosted Tree (GBT) Regressor, que é poderoso para capturar relações não-lineares nos dados.

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import GBTRegressor
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.sql.functions import col, hour, dayofweek

# ===============================
# 1. Preparação dos dados
# ===============================
df_gira_model = (
    gira_clean
    .filter(
        col("data").isNotNull() &
        col("num_bicicletas").isNotNull()
    )
    .withColumn("hora", hour(col("data")))
    .withColumn("dia_semana", dayofweek(col("data")))
)

print("Total de linhas para treino:", df_gira_model.count())

# ===============================
# 2. Usar TODOS os dados para treino
# ===============================
train_data = df_gira_model

# ===============================
# 3. Pipeline de Machine Learning
# ===============================
assembler = VectorAssembler(
    inputCols=["hora", "dia_semana"],
    outputCol="features"
)

gbt = GBTRegressor(
    featuresCol="features",
    labelCol="num_bicicletas",
    maxIter=10,
    seed=42
)

pipeline_gira = Pipeline(stages=[
    assembler,
    gbt
])

# ===============================
# 4. Treino do modelo
# ===============================
model_gira = pipeline_gira.fit(train_data)

print("Modelo treinado com sucesso")

# ===============================
# 5. Previsões
# ===============================
predictions = model_gira.transform(train_data)

predictions.select(
    "hora",
    "dia_semana",
    "num_bicicletas",
    "prediction"
).show(10)

# ===============================
# 6. Avaliação (RMSE)
# ===============================
evaluator = RegressionEvaluator(
    labelCol="num_bicicletas",
    predictionCol="prediction",
    metricName="rmse"
)

# 6. Avaliação do Modelo
evaluator_rmse = RegressionEvaluator(labelCol="num_bicicletas", predictionCol="prediction", metricName="rmse")
rmse = evaluator_rmse.evaluate(predictions)

evaluator_r2 = RegressionEvaluator(labelCol="num_bicicletas", predictionCol="prediction", metricName="r2")
r2 = evaluator_r2.evaluate(predictions)

print(f"Root Mean Squared Error (RMSE) nos dados de teste = {rmse:.2f}")
print(f"R² nos dados de teste = {r2:.2f}")
```

Previsões: A tabela de previsões mostra o número real de bicicletas (num_bicicletas) ao lado do valor previsto pelo modelo (prediction). Vemos que o modelo consegue aproximar-se dos valores reais.
RMSE (Root Mean Squared Error): Esta métrica indica o erro médio do modelo, na mesma unidade da variável alvo. Um RMSE de, por exemplo, 2.5 significa que, em média, as previsões do modelo erram por cerca de 2.5 bicicletas. Quanto menor o RMSE, melhor o modelo.
R² (R-squared): Indica a proporção da variância na disponibilidade de bicicletas que é explicada pelo nosso modelo. Um R² de 0.65, por exemplo, significa que 65% da "variação" no número de bicicletas pode ser explicada pelas features que usámos (estação, hora, dia da semana). Quanto mais próximo de 1, melhor.
Aplicação Prática: Este modelo pode ser integrado num dashboard para a equipa de logística da GIRA, que pode ver previsões de estações que ficarão vazias ou cheias nas próximas horas e redistribuir as bicicletas de forma proativa.


7.2 Previsão de Probabilidade de Congestionamento (Classificação)
O objetivo é prever a probabilidade de ocorrer um "evento de congestionamento". Primeiro, precisamos transformar o nosso problema num formato de classificação binária: definimos o que é um congestionamento (label = 1) e o que não é (label = 0). Uma abordagem simples é definir um "congestionamento" como qualquer local/hora que tenha mais do que um certo número de incidentes.
Depois, usaremos KMeans para agrupar coordenadas em zonas geográficas e um modelo de Regressão Logística para classificar se uma zona, numa determinada hora, terá ou não um congestionamento.

```python
df_cong_debug = (
    transito_clean
    .withColumn("lon", col("position.coordinates")[0])
    .withColumn("lat", col("position.coordinates")[1])
    .withColumn("hora", hour(col("creation_date_ts")))
    .withColumn("dia_semana", dayofweek(col("creation_date_ts")))
)

df_cong_debug.select(
    "lat", "lon", "hora", "dia_semana"
).summary("count").show()


from pyspark.ml.clustering import KMeans
from pyspark.ml.feature import VectorAssembler
from pyspark.sql.functions import col, hour, dayofweek
from pyspark.ml import Pipeline

# ===============================
# 1. Preparação dos dados
# ===============================

df_cong_ml = (
    transito_clean
    .withColumn("lon", col("position.coordinates")[0])
    .withColumn("lat", col("position.coordinates")[1])
    .withColumn("hora", hour(col("creation_date_ts")))
    .withColumn("dia_semana", dayofweek(col("creation_date_ts")))
    .filter(
        col("lat").isNotNull() &
        col("lon").isNotNull() &
        col("hora").isNotNull()
    )
)

print("Total de registos para ML:", df_cong_ml.count())

# ===============================
# 2. Vectorização
# ===============================

assembler = VectorAssembler(
    inputCols=["lat", "lon", "hora", "dia_semana"],
    outputCol="features"
)

# ===============================
# 3. KMeans
# ===============================

kmeans = KMeans(
    k=8,                  # número de zonas de congestionamento
    seed=42,
    featuresCol="features",
    predictionCol="cluster"
)

pipeline = Pipeline(stages=[assembler, kmeans])

# ===============================
# 4. Treino do modelo
# ===============================

model_cong = pipeline.fit(df_cong_ml)

# ===============================
# 5. Resultados
# ===============================

df_clusters = model_cong.transform(df_cong_ml)

df_clusters.select(
    "lat", "lon", "hora", "dia_semana", "cluster"
).show(10, truncate=False)


# 5. Avaliação do Modelo
evaluator_auc = BinaryClassificationEvaluator(labelCol="label", rawPredictionCol="rawPrediction", metricName="areaUnderROC")
auc = evaluator_auc.evaluate(predictions_cong)

# Accuracy
accuracy = predictions_cong.filter(predictions_cong.label == predictions_cong.prediction).count() / float(predictions_cong.count())


print(f"Area Under ROC (AUC) nos dados de teste = {auc:.2f}")
print(f"Acurácia nos dados de teste = {accuracy:.2f}")
```

Previsões: A saída mostra a probability (um vetor com a probabilidade de ser [classe 0, classe 1]) e a prediction final (0 ou 1). Uma probabilidade de [0.1, 0.9] indica uma chance de 90% de ser um evento de congestionamento.
AUC (Area Under ROC Curve): Esta é uma métrica excelente para modelos de classificação desbalanceados. Um valor de 0.5 é aleatório, enquanto 1.0 é um classificador perfeito. Um AUC de, por exemplo, 0.85 indica que o modelo tem uma ótima capacidade de distinguir entre uma situação de "congestionamento" e uma de "não-congestionamento".
Acurácia: Mede a percentagem de previsões corretas. É uma métrica útil, mas pode ser enganadora se as classes forem muito desbalanceadas (ex: se 99% dos casos não forem congestionamento, um modelo que prevê sempre "não congestionamento" terá 99% de acurácia, mas será inútil).
