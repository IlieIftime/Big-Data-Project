```python
from pyspark.sql import SparkSession

# Substitua se quiser outra versão do conector; exemplo de package para Spark 3.x e Scala 2.12
# spark-submit --packages org.mongodb.spark:mongo-spark-connector_2.12:10.1.0 your_script.py

uri = "mongodb+srv://bigdata:bigdata@projetobd.3lgjnzn.mongodb.net/test?retryWrites=true&w=majority"
database = "ProjetoBD"

spark = SparkSession.builder \
    .appName("MongoDB + PySpark Example") \
    .config("spark.mongodb.read.connection.uri", uri) \
    .config("spark.mongodb.write.connection.uri", uri) \
    .getOrCreate()

# Leitura da coleção 'condicionamentos'
df_cond = spark.read \
    .format("mongodb") \
    .option("connection.uri", uri) \
    .option("database", database) \
    .option("collection", "condicionamentos") \
    .load()

# Leitura da coleção 'mobilidade_lisboa'
df_mob = spark.read \
    .format("mongodb") \
    .option("connection.uri", uri) \
    .option("database", database) \
    .option("collection", "mobilidade_lisboa") \
    .load()

print("=== 5 primeiros documentos de 'condicionamentos' ===")
df_cond.show(5, truncate=False)

print("=== 5 primeiros documentos de 'mobilidade_lisboa' ===")
df_mob.show(5, truncate=False)
```

```python
df_cond.show(5, truncate=False)
```

```python
from pyspark.sql.functions import col, to_timestamp

gira = (
    df_mob
    .withColumn("timestamp", to_timestamp(col("entity_ts")))
    .select(
        col("desigcomercial").alias("estacao"),
        col("timestamp"),
        col("numbicicletas").alias("bicicletas"),
        col("numdocasvacias").alias("docas_livres")
    )
)

gira.show(5)


transito = (
    df_cond
    .withColumn("timestamp", to_timestamp(col("creation_date")))
    .select(
        col("timestamp"),
        col("impacto"),
        col("motivo"),
        col("morada")
    )
)

transito.show(5)
```

# Limpeza e preparação dos dados

Converte datas de texto para timestamp (Spark só trabalha bem assim)
Remove registos inválidos (datas vazias ou bicicletas negativas)
Seleciona apenas as colunas relevantes
Transforma dados “crus” em dados prontos para análise

```python
from pyspark.sql.functions import col, to_timestamp

gira_clean = (
    df_mob
    .withColumn("timestamp", to_timestamp(col("entity_ts")))
    .dropna(subset=["timestamp"])
    .filter(col("numbicicletas") >= 0)
    .select(
        col("desigcomercial").alias("estacao"),
        col("timestamp"),
        col("numbicicletas").alias("bicicletas"),
        col("numdocasvacias").alias("docas_livres")
    )
)

gira_clean.show(5)

transito_clean = (
    df_cond
    .withColumn("timestamp", to_timestamp(col("creation_date")))
    .dropna(subset=["timestamp"])
    .select(
        col("timestamp"),
        col("impacto"),
        col("motivo"),
        col("morada")
    )
)

import unicodedata
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

def normalizar_texto(texto):
    if texto is None:
        return None
    texto = unicodedata.normalize("NFKD", texto)
    texto = texto.encode("ASCII", "ignore").decode("ASCII")
    return texto

normalizar_udf = udf(normalizar_texto, StringType())

gira_clean_norm = gira_clean.withColumn(
    "estacao_normalizada",
    normalizar_udf("estacao")
)

transito_clean.show(5)
```

Foi utilizada a média diária do número de bicicletas disponíveis para representar o comportamento geral do sistema GIRA em cada dia e permitir a comparação com os condicionamentos de trânsito
A média diária foi usada para resumir vários registos do mesmo dia num único valor representativo.

```python
from pyspark.sql.functions import to_date, avg, count

gira_por_dia = (
    gira_clean
    .withColumn("data", to_date(col("timestamp")))
    .groupBy("data")
    .agg(avg("bicicletas").alias("media_bicicletas"))
    .orderBy("data")
)

gira_por_dia.show(5)

transito_por_dia = (
    transito_clean
    .withColumn("data", to_date(col("timestamp")))
    .groupBy("data")
    .agg(count("*").alias("num_condicionamentos"))
    .orderBy("data")
)

transito_por_dia.show(5)
```

# Integração dos datasets

```python
df_integrado = gira_por_dia.join(
    transito_por_dia,
    on="data",
    how="inner"
)

df_integrado.show(10)
```

# Análise exploratória

### Gráfico 1 — Evolução diária da disponibilidade média de bicicletas

```python
from pyspark.sql.functions import to_date, avg

gira_bicicletas_dia = (
    gira_clean
    .withColumn("data", to_date("timestamp"))
    .groupBy("data")
    .agg(avg("bicicletas").alias("media_bicicletas"))
    .orderBy("data")
)

df_plot = gira_bicicletas_dia.toPandas()

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(df_plot["data"], df_plot["media_bicicletas"])
plt.xlabel("Data")
plt.ylabel("Média de bicicletas")
plt.title("Evolução diária da disponibilidade média de bicicletas (GIRA)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Gráfico 2 — Estações com menor disponibilidade média de bicicletas

```python
from pyspark.sql.functions import avg

estacoes_todas = (
    gira_clean_norm
    .groupBy("estacao_normalizada")
    .agg(avg("bicicletas").alias("media_bicicletas"))
    .orderBy("media_bicicletas")
)

df_plot = estacoes_todas.toPandas()

import matplotlib.pyplot as plt

plt.figure(figsize=(14,6))
plt.bar(df_plot["estacao_normalizada"], df_plot["media_bicicletas"])
plt.xlabel("Estação")
plt.ylabel("Média de bicicletas")
plt.title("Disponibilidade média de bicicletas por estação (GIRA)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
